# Automatic Mask Generator

## Libraries

```python
import torch
import torchvision
print(torch.__version__, torch.cuda.is_available())
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

import sys
sys.path.append('segment-anything')
```

```python
from segment_anything import sam_model_registry, SamAutomaticMaskGenerator, SamPredictor
```

## Implementation

```python
image = cv2.imread('imgs/brain_tumor.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 10))
plt.imshow(image)
plt.axis('off')
plt.show()
```

```python
sam_checkpoint = 'sam_vit_h_4b8939.pth'
model_type = 'vit_h'

device = 'cuda'
```

```python
sam = sam_model_registry[model_type](checkpoint = sam_checkpoint)
sam.to(device)
```

```python
mask_generator_ = SamAutomaticMaskGenerator(
    model = sam,
    points_per_side = 32,
    pred_iou_thresh = 0.9, # include only masks with iou > 0.9
    stability_score_thresh = 0.96, # include only stable masks
    crop_n_layers  = 1,
    min_mask_region_area = 10,
)

masks = mask_generator_.generate(image)
```

```python
masks
```

```python
print(len(masks))
```

```python
def show_anns(anns):
    if len(anns) == 0:
        return
    sorted_anns = sorted(anns, key=(lambda x: x['area']), reverse=True)
    ax = plt.gca()
    ax.set_autoscale_on(False)

    img = np.ones((sorted_anns[0]['segmentation'].shape[0], sorted_anns[0]['segmentation'].shape[1], 4))
    img[:,:,3] = 0
    for ann in sorted_anns:
        m = ann['segmentation']
        color_mask = np.concatenate([np.random.random(3), [0.35]])
        img[m] = color_mask
    ax.imshow(img)
```

```python
plt.figure(figsize=(10,10))
plt.imshow(image)
show_anns(masks)
plt.axis('off')
plt.show() 
```

```python
mask_generator_tuned = SamAutomaticMaskGenerator(
    model = sam,
    points_per_side = 32,
    pred_iou_thresh = 0.9, # include only masks with iou > 0.9
    stability_score_thresh = 0.96, # include only stable masks
    crop_n_layers  = 1,
    min_mask_region_area = 10,
)

masks_2 = mask_generator_tuned.generate(image)
```

```python
masks_2
```



Mask generation returns a list over masks, where each mask is a dictionary containing various data about the mask. These keys are:

- `segmentation` : the mask
- `area` : the area of the mask in pixels
- `bbox` : the boundary box of the mask in XYWH format
- `predicted_iou` : the model's own prediction for the quality of the mask
- `point_coords` : the sampled input point that generated this mask
- `stability_score` : an additional measure of mask quality
- `crop_box` : the crop of the image used to generate this mask in XYWH format



```python
plt.figure(figsize=(10,10))
plt.imshow(image)
show_anns(masks_2)
plt.axis('off')
plt.show() 
```

# Promptable Mask Predictor

The `SamPredictor` class provides an easy interface to the model for prompting the model. It allows the user to first set an image using the `set_image` method, which calculates the necessary image embeddings. Then, prompts can be provided via the `predict` method to efficiently predict masks from those prompts. The model can take as input both point and box prompts, as well as masks from the previous iteration of prediction.

## Set-up

```python
import numpy as np
import torch
import matplotlib.pyplot as plt
import cv2
```

```python
def show_mask(mask, ax, random_color=False):
    if random_color:
        color = np.concatenate([np.random.random(3), np.array([0.6])], axis=0)
    else:
        color = np.array([30/255, 144/255, 255/255, 0.6])
    h, w = mask.shape[-2:]
    mask_image = mask.reshape(h, w, 1) * color.reshape(1, 1, -1)
    ax.imshow(mask_image)
    
def show_points(coords, labels, ax, marker_size=375):
    pos_points = coords[labels==1]
    neg_points = coords[labels==0]
    ax.scatter(pos_points[:, 0], pos_points[:, 1], color='green', marker='*', s=marker_size, edgecolor='white', linewidth=1.25)
    ax.scatter(neg_points[:, 0], neg_points[:, 1], color='red', marker='*', s=marker_size, edgecolor='white', linewidth=1.25)   
    
def show_box(box, ax):
    x0, y0 = box[0], box[1]
    w, h = box[2] - box[0], box[3] - box[1]
    ax.add_patch(plt.Rectangle((x0, y0), w, h, edgecolor='green', facecolor=(0,0,0,0), lw=2))    
```

```python
image = cv2.imread('imgs/brain_tumor.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(10,10))
plt.imshow(image)
plt.axis('on')
plt.show()
```

## Prompting

### One point specification

```python
import sys
sys.path.append('segment-anything')
from segment_anything import sam_model_registry, SamPredictor

sam_checkpoint = "sam_vit_h_4b8939.pth"
model_type = "vit_h"

device = "cuda"

sam = sam_model_registry[model_type](checkpoint=sam_checkpoint)
sam.to(device=device)

predictor = SamPredictor(sam)
```

process the image and produce image embedding by calling `SamPredictor.set_image`.`SamPredictor` remembers this embedding and will use it for subsequent mask prediction.

```python
predictor.set_image(image)
```

```python
input_point = np.array([[550, 370]])
input_label = np.array([1])
```

```python
plt.figure(figsize=(10,10))
plt.imshow(image)
show_points(input_point, input_label, plt.gca())
plt.axis('on')
plt.show()  
```

Predict with `SamPredictor.predict`. The model returns masks, quality predictions for those masks, and low resolution mask logits that can be passed to the next iteration of prediction.

```python
masks, scores, logits = predictor.predict(
    point_coords=input_point,
    point_labels=input_label,
    multimask_output=True,
)
masks_2, scores_2, logits_2 = predictor.predict(
    point_coords=input_point,
    point_labels=input_label,
    multimask_output=False,
)
```

we use `multimask_outout=True`(default setting) to diplay the3 masks wach one with its score, it is intended for ambiguous prompts. If its is `False` it does not return the mask with the highest score.

```python
masks.shape, masks_2.shape  # (number_of_masks) x H x W
```

```python
for i, (mask, score) in enumerate(zip(masks, scores)):
    plt.figure(figsize=(10,10))
    plt.imshow(image)
    show_mask(mask, plt.gca())
    show_points(input_point, input_label, plt.gca())
    plt.title(f"Mask {i+1}, Score: {score:.3f}", fontsize=18)
    plt.axis('off')
    plt.show()  
  
```

```python
for i, (mask, score) in enumerate(zip(masks_2, scores_2)):
    plt.figure(figsize=(10,10))
    plt.imshow(image)
    show_mask(mask, plt.gca())
    show_points(input_point, input_label, plt.gca())
    plt.title(f"Mask {i+1}, Score: {score:.3f}", fontsize=18)
    plt.axis('off')
    plt.show()  
  
```

### Specifying a specific object with additional points


The single input point is ambiguous, and the model has returned multiple objects consistent with it. To obtain a single object, multiple points can be provided. If available, a mask from a previous iteration can also be supplied to the model to aid in prediction. When specifying a single object with multiple prompts, a single mask can be requested by setting `multimask_output=False`.

```python
input_point = np.array([[575, 380], [500, 350]])
input_label = np.array([1, 1])

mask_input = logits[np.argmax(scores), :, :]  # Choose the model's best mask
```

```python
masks, _, _ = predictor.predict(
    point_coords=input_point,
    point_labels=input_label,
    mask_input=mask_input[None, :, :],
    multimask_output=False,
)
```

```python
plt.figure(figsize=(10,10))
plt.imshow(image)
show_mask(masks, plt.gca())
show_points(input_point, input_label, plt.gca())
plt.axis('off')
plt.show() 
```

```python
masks, _, _ = predictor.predict(
    point_coords=input_point,
    point_labels=input_label,
    mask_input=mask_input[None, :, :],
    multimask_output=False,
)
```

```python
plt.figure(figsize=(10, 10))
plt.imshow(image)
show_mask(masks, plt.gca())
show_points(input_point, input_label, plt.gca())
plt.axis('off')
plt.show() 
```

### Specifiying with a bounding box

The model can also take a box as input, provided in xyxy format.

```python
input_box = np.array([420, 210, 610, 450])
```

```python
masks, _, _ = predictor.predict(
    point_coords=None,
    point_labels=None,
    box=input_box[None, :],
    multimask_output=False,
)
```

```python
plt.figure(figsize=(10, 10))
plt.imshow(image)
show_mask(masks[0], plt.gca())
show_box(input_box, plt.gca())
plt.axis('on')
plt.show()
```

## Combining points and boxes

Points and boxes may be combined, just by including both types of prompts to the predictor. Here this can be used to select just the trucks's tire, instead of the entire wheel.

```python
input_box = np.array([420, 210, 610, 450])
input_point = np.array([[575, 300]])
input_label = np.array([0])
```

```python
masks, _, _ = predictor.predict(
    point_coords=input_point,
    point_labels=input_label,
    box=input_box,
    multimask_output=False,
)
```

```python
plt.figure(figsize=(10, 10))
plt.imshow(image)
show_mask(masks[0], plt.gca())
show_box(input_box, plt.gca())
show_points(input_point, input_label, plt.gca())
plt.axis('off')
plt.show()
```
